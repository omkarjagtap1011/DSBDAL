```python
# To import built-in titanic dataset
import seaborn as sns

# For data analysis
import pandas as pd

# Importing matplotlib for visualization
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
# No. of rows and cols (dimension)
df.shape
```

```python
# Basic Information of dataset
df.info()
```

```python
# Get statistical analysis
df.describe()
```

```python
# Count plot for survived column
# To check how many people survived
sns.countplot(x="survived", data=df)
plt.title("Survival Count")
plt.show()

# Observations:-

# More passengers did not survive than survived
# Survival rate is clearly below 50%
```

```python
# Survival by gender
# To check gender vise survival
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Survival by Gender")
plt.show()

# Women survived at a much higher rate than men
# One of the strongest signals in the dataset
# "Women and children first" pattern can be seen
```

```python
# Survival by passenger class
# To check the priority
sns.countplot(x="pclass", hue="survived", data=df)
plt.title("Survival by Passenger Class")
plt.show()

# Observation:-
# Class 1 Passengers have highest survival rate
# Class 3 Passengers have lowest survival rate
```

```python
# Age distribution and survival
sns.histplot(data=df, x="age", hue="survived", bins=30, kde=True)
plt.title("Age Distribution by Survival")
plt.show()

# Observation:-
# Children had higher survival rates
# Older adults were less likely to survive
# Overlap exists, but the trend is visible
```

```python
# Plot histogram of fare distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="fare", bins=40, kde=True)

plt.title("Distribution of Titanic Ticket Prices (Fare)")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()

# Observation:-
# The distribution is right-skewed
# Most passengers paid low fares
# A small number paid very high prices
```

```python
# Fare vs survival
# To check survival rate as per person's wealth

sns.boxplot(x="survived", y="fare", data=df)
plt.title("Fare Paid vs Survival")
plt.show()

# Observation:-

# Passengers who did NOT survive (survived = 0)
# Lower median fare
# Most fares are concentrated at the cheaper end
# Fewer high-value outliers

# Passengers who survived (survived = 1)
# Higher median fare
# Wider spread of fares
# Many high-fare outliers
```

```python
#  Plot a box plot for distribution of age with respect to each gender along with the information about whether they survived or not.
# Columns used will be 'sex' and 'age'

sns.boxplot(x='sex', y='age', hue='survived', data=df)
plt.title("Age per gender vs survival")
plt.show()

# Observations:-
```
