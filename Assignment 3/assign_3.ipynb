```python
df = pd.read_csv('iris.csv')
df = df.drop('Id', axis=1)
```

```python
cols = list(df.columns)
cols.remove('Species')

for col in cols:
    mean = df.groupby('Species')[col].mean()
    print(f"Mean for {col}:")
    print(mean)
    print()

    std = df.groupby('Species')[col].std()
    print(f"Std dev for {col}:")
    print(std)
    print()

    qtl_25 = df.groupby('Species')[col].quantile(0.25)
    print(f"25% for {col}:")
    print(qtl_25)
    print()
```
