```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Assign10_Iris.csv")
df.head()
```

```python
X = df.drop(columns=['Id', 'Species'])
y = df['Species']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=24, shuffle=True)
```

```python
class NaiveBayes:
    def fit(self, X_train, y_train):
        self.classes = np.unique(y_train)

        self.prior = [len(y_train[y_train==c]) / len(y_train) for c in self.classes]

        self.means = [X_train[y_train==c].mean() for c in self.classes]
        self.stds = [X_train[y_train==c].std() for c in self.classes]
    
    def compute_liklihood(self, row, class_index):
        likelihood = 1

        for feature in row.index:
            mean = self.means[class_index][feature]
            std = self.stds[class_index][feature]

            likelihood *= (1/(std * np.sqrt(2 * np.pi))) * np.exp(-(((row[feature] - mean)**2)/(2 * (std)**2)))
        
        return likelihood
    
    def predict(self, X):
        y_pred = []
        for _, row in X.iterrows():
            posterior = []
            for i in range(len(self.classes)):
                likelihood = self.compute_liklihood(row, i)
                posterior.append(likelihood * self.prior[i])
            
            y_pred.append(self.classes[np.argmax(posterior)])
        return np.array(y_pred)


```

```python
nb = NaiveBayes()
nb.fit(X_train, y_train)
y_pred = nb.predict(X_test)
```

```python
acc = np.mean(y_pred==y_test) * 100
```
