```python
# for data visualization
import pandas as pd

import seaborn as sns

import matplotlib.pyplot as plt
```

```python
# load iris dataset into dataframe
df = pd.read_csv('iris.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
# 1. Listing features and their types

df.info()
```

```python
# 2. create histogram for each feature in the dataset to illustrate feature distribution
# This operation can be done only on numeric features

cols = df.columns.to_list()
cols.remove('Id')
cols.remove('Species')

for col in cols:
    plt.title(f"Distribution of {col}")
    sns.histplot(data=df, x=col, kde=True, bins=20)
    plt.show()

# Observations:-
# SepalLengthCm and SepalWidthCm is normally distributed
# PetalLengthCm and PetalWidthCm have outliers on the left
```

```python
# Using boxplot on all the numeric data
sns.boxplot(data=df.drop(['Species', 'Id'], axis=1))

# After observing the boxplot, I can observe SepalWidthCm contains
```

```python
# Using boxplot for each species per column
for col in cols:
    plt.title(f"Boxplot for {col}")
    sns.boxplot(data=df, x="Species", y=col)
    plt.show()
```

```python
# Observation:-
# Setosa is compact
# versicolor and virginica have more spread
# Identified the outliers
```
