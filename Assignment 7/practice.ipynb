```python
import nltk
import string
import math
```

```python
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
with open('data.txt', 'r') as f:
    text = f.read()

alphabet = string.printable

def clean(text):
    return ''.join([word for word in text.lower() if word in alphabet])

text = clean(text)
print(text)
```

```python
tokens = word_tokenize(text)
print(tokens)
```

```python
pos_tags=nltk.pos_tag(tokens)
print(pos_tags)
```

```python
stop_words = set(stopwords.words('english'))
removed_tokens = []
filtered_tokens = []

for word in tokens:
    filtered_tokens.append(word) if word not in stop_words else removed_tokens.append(word)
        

filtered_tokens = [word for word in filtered_tokens if word.lower() not in string.punctuation]
print(filtered_tokens)
print(removed_tokens)
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_tokens]
stemmed_words
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]

lemmatized_words
```

TF(t, d) = (total no. of times term t appeard in document d)/(Total no. of terms in document d)

It measures how important the word is across entire corpus
IDF(t) = log((total no. of documents)/(no. documents containing term t))


```python
documentA = "Hii, my name is Omkar"
documentB = "I am a student"
```

```python
bagOfWordsA = documentA.split(" ")
bagOfWordsB = documentB.split(" ")
```

```python
# Corpus
uniqueWords = set(bagOfWordsA).union(set(bagOfWordsB))
```

```python
# Creating frequency dictionary for each unique word for each document

freqA = dict.fromkeys(uniqueWords, 0)
for word in bagOfWordsA:
    freqA[word]+=1

freqB = dict.fromkeys(uniqueWords, 0)
for word in bagOfWordsB:
    freqB[word]+=1
```

```python
def tfCalculator(wordDict, document):
    tfDict = {}
    N = len(document)

    for word, count in wordDict.items():
        tfDict[word] = (count/float(N))
    
    return tfDict

tfA = tfCalculator(freqA, bagOfWordsA)
tfB = tfCalculator(freqB, bagOfWordsB)
```

```python
import math
def idfCalculator(documents):
    N = len(documents)

    idf = dict.fromkeys(uniqueWords, 0)

    for word in uniqueWords:
        for document in documents:
            if word in document:
                idf[word]+=1
                break


    for word, count in idf.items():
        idf[word] = math.log((N)/(count))
    
    return idf

idf = idfCalculator([bagOfWordsA, bagOfWordsB])
idf
```

```python
def tfidfCalculator(tf, idf):
    tfidf = {}

    for word, val in tf.items():
        tfidf[word] = val * idf[word]
    
    return tfidf

tfidfA = tfidfCalculator(tfA, idf)
tfidfB = tfidfCalculator(tfB, idf)
```
